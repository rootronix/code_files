<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Explain repetition using an everyday task.</div>

Imagine calling three students for attendance. Separate print statements work, but every new student needs another line. A loop lets us write the repeated action once and keep the names in a list.


#### Example 1 — Repeating the action ourselves

*Predict the output, then run the code cell below.*


```python
print("Hello", "Arun")
print("Hello", "Bina")
print("Hello", "Carl")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The action is the same on all three lines. Only the name changes.</div>


#### Example 2 — Letting a loop repeat the action

*Predict the output, then run the code cell below.*


```python
names = ["Arun", "Bina", "Carl"]
for name in names:
    print("Hello", name)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><ol style="margin:8px 0 0;padding-left:24px;"><li>Python takes the first name.</li><li>It runs the indented print.</li><li>It repeats for each remaining name and stops when the list ends.</li></ol></div>


#### Three words we will use

| Word | Meaning in this example |
| --- | --- |
| Loop | The code that repeats the greeting. |
| Iteration | One pass through the loop body; one greeting. |
| Iterable | Something Python can get items from one at a time; names is an iterable. |

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">2. Read a for loop line by line</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Identify the loop variable, colon, and indented body.</div>


#### Example 1 — Inside and outside the body

*Predict the output, then run the code cell below.*


```python
for fruit in ["apple", "banana"]:
    print("Wash", fruit)
    print("Pack", fruit)
print("All packed")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>for begins the loop. fruit is a variable that receives the current item. in connects it to the source. The colon starts a block. Both indented lines run for each fruit. The final unindented line runs once after the loop.</div>

| Pass | fruit | Actions |
| --- | --- | --- |
| 1 | "apple" | Wash apple, then pack apple. |
| 2 | "banana" | Wash banana, then pack banana. |
| After the loop | "banana" | Print All packed once. |


#### Example 2 — The loop variable after a nonempty loop

*Predict the output, then run the code cell below.*


```python
for number in [10, 20, 30]:
    print(number)
print("Last:", number)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>number is an ordinary variable. Each pass assigns the next value to it. Because the loop ran, it ends with 30.</div>


#### Example 3 — An empty iterable

*Predict the output, then run the code cell below.*


```python
item = "unchanged"
for item in []:
    print("Inside")
print(item)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The body runs zero times. The loop makes no assignment to item, so its earlier value stays. If item had never been defined, reading it afterward would raise <code>NameError</code>.</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">3. Loop through familiar collections</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Connect loops to the earlier dictionaries lesson.</div>

The loop structure stays the same when the source changes. A string provides characters. Lists and tuples provide their items. A dictionary provides keys unless we ask for values or pairs.


#### Example 1 — A string and a tuple

*Predict the output, then run the code cell below.*


```python
for l in "Hi":
    print(l)
for day in ("Mon", "Tue"):
    print(day)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The first loop visits two characters. The second visits two complete strings; each tuple item is one day.</div>


#### Example 2 — Dictionary keys and values

*Predict the output, then run the code cell below.*


```python
scores = {"Arun": 82, "Bina": 95}
for name in scores:
    print("Name:", name)
for score in scores.values():
    print("Score:", score)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>for name in scores reads the keys. scores.<code>values()</code> supplies just the values. Both follow the dictionary insertion order.</div>


#### Example 3 — Dictionary pairs

*Predict the output, then run the code cell below.*


```python
scores = {"Arun": 82, "Bina": 95}
for key, value in scores.items():
    print(key, "scored", value)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><code>items()</code> supplies one pair at a time, such as (&quot;Arun&quot;, 82). Unpacking assigns its first item to name and its second item to score.</div>

| Source | What the loop receives |
| --- | --- |
| List or tuple | Each item, in sequence order. |
| String | Each character, including spaces and punctuation. |
| Dictionary | Each key, in insertion order. |
| Set | Each unique item; no guaranteed iteration order. |

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">4. Count with range</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Choose a start, an excluded stop, and a step.</div>

`range()` represents a sequence of integers. Use it when the task needs number values or a fixed number of repetitions. It is not a list; list(`range(...)`) is useful when you want to display all its values.


#### Example 1 — Start at zero

*Predict the output, then run the code cell below.*


```python
for number in range(5):
    print(number)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><code>range(3)</code> starts at 0, uses a step of 1, and stops before 3. The body runs three times.</div>


#### Example 2 — Choose the first number

*Predict the output, then run the code cell below.*


```python
for page_number in range(1, 4):
    print("Page", page_number)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The start, 1, is included. The stop, 4, is excluded. To include 3 when counting upward by 1, use stop=4.</div>


#### Example 3 — Step forward or backward

*Predict the output, then run the code cell below.*


```python
print(list(range(2, 9, 2)))
print(list(range(5, 0, -1)))
print(list(range(5, 0)))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>A step of 2 adds two each time. A step of -1 subtracts one. The last range is empty because its default positive step cannot move from 5 toward numbers below 0.</div>

| Form | Values |
| --- | --- |
| `range(4)` | 0, 1, 2, 3 |
| `range(2, 5)` | 2, 3, 4 |
| `range(1, 7, 2)` | 1, 3, 5 |
| `range(4, -1, -1)` | 4, 3, 2, 1, 0 |
| `range(0)` or `range(3, 3)` | No values |

Watch out: a step of zero raises ValueError. Choose a positive or negative nonzero step. All range arguments must be integers (or integer-like objects), not decimal values such as 1.5.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Use if to choose an action for each item.</div>

A loop selects the next item; an if statement decides what to do with that item. The if body is one indentation level deeper than the loop body. Here, % is the remainder operator: number % 2 == 0 means number is even.


#### Example 1 — Print the even numbers

*Predict the output, then run the code cell below.*


```python
for number in range(1, 7):
    if number % 2 == 0:
        print(number, "is even")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><ol style="margin:8px 0 0;padding-left:24px;"><li>Visit each number from 1 through 6.</li><li>Check its remainder after division by 2.</li><li>Print only when that remainder is zero.</li></ol></div>


#### Example 2 — Report a result for every student

*Predict the output, then run the code cell below.*


```python
scores = {"Arun": 82, "Bina": 45, "Carl": 50}
for name, score in scores.items():
    if score >= 50:
        print(name, "Pass")
    else:
        print(name, "Needs practice")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The threshold includes 50. Each student takes exactly one branch. This else belongs to if because they have the same indentation.</div>


#### Example 3 — Find low stock

*Predict the output, then run the code cell below.*


```python
stock = {"Pen": 8, "Book": 2, "Bag": 0}
for item, quantity in stock.items():
    if quantity < 3:
        print("Order more:", item)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The loop visits every product. Only products with a quantity below 3 are printed.</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">6. Build totals and counts</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Initialise a result once and update it inside the loop.</div>

An accumulator stores a result that grows as the loop runs. Put the starting value before the loop. total += price means total = total + price. A total adds item values; a count adds 1 for each matching item.


#### Example 1 — Add three prices

*Predict the output, then run the code cell below.*


```python
prices = [20, 30, 10]
total = 0
for price in prices:
    total = total + price
print("Total:", total)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>Start with 0 because adding zero changes nothing. Add each price to the previous total. Print the final total after the loop.</div>

| Pass | price | total before | total after |
| --- | --- | --- | --- |
| 1 | 20 | 0 | 20 |
| 2 | 30 | 20 | 50 |
| 3 | 10 | 50 | 60 |


#### Example 2 — Count passing scores

*Predict the output, then run the code cell below.*


```python
scores = [82, 45, 50, 91]
pass_count = 0
for score in scores:
    if score >= 50:
        pass_count += 1
print("Passed:", pass_count)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>Increase the count by 1 for 82, 50, and 91. Adding score instead would calculate a total, not a count.</div>


#### Example 3 — Calculate an average safely

*Predict the output, then run the code cell below.*


```python
scores = [60, 80, 100]
total = 0
for score in scores:
    total += score
print("Total:", total)
if len(scores) > 0:
    print("Average:", total / len(scores))
else:
    print("No scores")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><code>len(scores)</code> gives the number of scores. The empty-list check prevents division by zero. Try scores = [] and the message becomes No scores.</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">7. Collect results and track the largest value</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Build a new list or keep the best value seen so far.</div>


#### Example 1 — Save matching items

*Predict the output, then run the code cell below.*


```python
words = ["sky", "tree", "go", "river"]
long_words = []
for word in words:
    print(word)
    if len(word) >= 4:
        long_words.append(word)
    print(long_words)

print(long_words)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><ol style="margin:8px 0 0;padding-left:24px;"><li>Create one empty list.</li><li>Check each word.</li><li><code>append()</code> adds each match to the end of the result. The original words list stays unchanged.</li></ol></div>


#### Example 2 — Transform each item and join text

*Predict the output, then run the code cell below.*


```python
names = ["Arun", "Bina", "Carl"]
labels = []
for name in names:
    labels.append("Hello " + name)
print("; ".join(labels))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>Each iteration builds a greeting and saves it. <code>join()</code> combines the strings once, with a semicolon and space between them.</div>


#### Example 3 — Find a maximum including negative numbers

*Predict the output, then run the code cell below.*


```python
readings = [-8, -3, -11]
if len(readings) == 0:
    print("No readings")
else:
    best = readings[0]
    for reading in readings:
        if reading > best:
            best = reading
    print("Largest:", best)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>Check for empty input before reading position 0. Start best with a real value. It stays -8, becomes -3, then stays -3. Starting at 0 would give a wrong answer for these negative readings.</div>

```python
for number in [4, 8, 15, 20]:
    if number > 10:
        print("First over 10:", number)
        break
    print("Checked:", number)
print("Search complete")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>4 and 8 do not match. At 15, print the result and break. The loop never visits 20.</div>

#### Example 2 — Update before continue in a while loop

*Predict the output, then run the code cell below.*


```python
number = 0
while number < 5:
    number += 1
    if number % 2 == 0:
        continue
    print(number)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The counter updates before the skip. #### Example 3 — pass performs no action

*Predict the output, then run the code cell below.*


```python
for number in [1, 2, 3]:
    if number == 2:
        pass
    print(number)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>pass is a placeholder where Python requires a statement. It does nothing, so the print still runs for 2. Replacing pass with continue would skip that print; replacing it with break would end the loop at 2.</div>

| Statement | Effect |

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Recognise a stop marker and handle its absence.</div>

A sentinel is a special value that means stop, such as the text "STOP". It should be chosen so it cannot be confused with normal data. Check for the sentinel before processing the item.


#### Example 1 — A marker in the data

*Predict the output, then run the code cell below.*


```python
messages = ["hello", "report", "STOP", "ignored"]
for message in messages:
    if message == "STOP":
        break
    print("Process:", message)
print("Done")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The first two messages are processed. STOP ends the loop, so the marker and the later item are not processed.</div>


#### Example 2 — The marker is missing

*Predict the output, then run the code cell below.*


```python
messages = ["hello", "report"]
index = 0
while index < len(messages) and messages[index] != "STOP":
    print("Process:", messages[index])
    index += 1
print("Done")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><ol style="margin:8px 0 0;padding-left:24px;"><li>Check the index is valid.</li><li>Only then read the item and test the marker.</li><li>Process and advance. and checks left to right and skips the second test when the first is <code>False</code>. This also handles an empty list.</li></ol></div>


#### Why the boundary check comes first

Reading messages[index] before checking index &lt; `len(messages)` can raise `IndexError` after the final item. The for version naturally ends with the source. Use the indexed while version only when position is part of the task.

Input validation means checking that a value is allowed before using it. For this demonstration, "1" means View and "2" means Exit. A list stands in for user input so the notebook runs without waiting for typing. These examples validate menu choices, not arbitrary numbers.


#### Example 1 — A valid choice arrives

*Predict the output, then run the code cell below.*


```python
attempts = ["hello", "9", "2"]
index = 0
choice = None
while index < len(attempts):
    candidate = attempts[index]
    index += 1
    if candidate in ("1", "2"):
        choice = candidate
        break
    print("Rejected:", candidate)
if choice is None:
    print("No valid choice")
else:
    print("Accepted:", choice)
```

#### Example 2 — All attempts are invalid

*Predict the output, then run the code cell below.*


```python
attempts = ["hello", "9"]
choice = None
for candidate in attempts:
    if candidate in ("1", "2"):
        choice = candidate
        break
    print("Rejected:", candidate)
if choice is None:
    print("No valid choice")
else:
    print("Accepted:", choice)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The for version naturally handles a finite sequence. No match means choice remains <code>None</code>. Both versions also accept an empty attempts list and report No valid choice.</div>

Use the same small inventory from the dictionaries lesson. Each product has a price and quantity. Assume prices and quantities are nonnegative numbers. We will build the report in three stages, with each stage able to run on its own.


#### Stage 1 Calculate the value of each product

*Predict the output, then run the code cell below.*


```python
inventory = {
    "Notebook": {"price": 80, "quantity": 3},
    "Backpack": {"price": 650, "quantity": 1},
    "Pen": {"price": 20, "quantity": 8},
}
for name, details in inventory.items():
    value = details["price"] * details["quantity"]
    print(name, value)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><ol style="margin:8px 0 0;padding-left:24px;"><li><code>items()</code> provides a product name and its details dictionary.</li><li>Read price and quantity from that dictionary.</li><li>Multiply them to calculate stock value for the product.</li></ol></div>


#### Stage 2 Add the product values

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">18. Common mistakes and a quick reference</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Recognise the symptom and choose a correction.</div>

| Mistake | What happens | Correction |
| --- | --- | --- |
| Missing colon or indentation | Python cannot identify the loop body. | Use a colon and indent the body four spaces. |
| Wrong range stop | The final wanted number is missing. | The stop is excluded; inspect list(`range(...)`). |
| Resetting a total inside the loop | Earlier additions are lost. | Initialise once before the loop. |
| Missing while update | The condition may remain `True` forever. | Trace how the condition eventually becomes `False` or a break is reached. |
| Update after continue | Some passes skip the update. | Move the essential update before the possible continue. |
| Using pass to skip | Later code still runs. | Use continue to skip the remaining body. |
| Reading list[0] on an empty list | `IndexError` occurs. | Check the list length first. |
| Starting a maximum at zero | All-negative input can give the wrong result. | Start from a real input value after an empty check. |
| Unequal sources in zip | Extra items have no output pair. | Check lengths when each item must be paired. |
| Adding or deleting during direct iteration | List items may be skipped; dict size changes can raise RuntimeError. | Build a new result or loop over a suitable copy. |


#### Pick the tool by the job

| Need | Tool |
| --- | --- |
| Each value | for value in values |
| A number sequence | `range(start, stop, step)` |
| A condition controls repetition | while condition |
| A position and value | `enumerate(values)` |
| Related values from separate sources | `zip(left, right)` |
| Dictionary keys and values | data.`items()` |
| Stop or skip | break or continue |
| A saved total or collection | An accumulator created before the loop |

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Recognise normal completion and an early break.</div>

A loop else belongs at the same indentation as for or while. It runs when the loop finishes normally: a for loop exhausts its source, or a while condition becomes `False`. A break skips it. An exception or a return also leaves the loop without running its else.


#### Example 1 — The search finds a match

*Predict the output, then run the code cell below.*


```python
for number in [2, 5, 8]:
    if number % 2 != 0:
        print("Found odd:", number)
        break
else:
    print("No odd number")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The loop reaches 5 and breaks. The else block is skipped. This else lines up with for, so it does not belong to the inner if.</div>


#### Example 2 — The search finishes without a match

*Predict the output, then run the code cell below.*


```python
for number in [2, 4, 8]:
    if number % 2 != 0:
        print("Found odd:", number)
        break
else:
    print("No odd number")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>No item triggers break. Exhausting the list runs else. An empty list also exhausts normally and runs else.</div>


#### Example 3 — A while loop finishes normally

*Predict the output, then run the code cell below.*


```python
count = 2
while count > 0:
    print(count)
    count -= 1
else:
    print("Countdown complete")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>When count becomes 0, the condition is <code>False</code> and else runs. If the initial count were 0, the body would run zero times and else would still run.</div>

#### Example 4 — A while loop stops early

*Predict the output, then run the code cell below.*


```python
count = 3
while count > 0:
    if count == 2:
        print("Stopped early")
        break
    print(count)
    count -= 1
else:
    print("Countdown complete")
print("After loop")
```

#### Trace the four possible cases

| Case | Does the body run | Does else run |
| --- | --- | --- |
| For source has no items | No | Yes |
| While condition starts `False` | No | Yes |
| Loop completes normally | Maybe; zero or more times | Yes |
| A break executes | Yes, up to break | No |

Use a flag when that is easier for the class to read. Loop else is especially useful for a search where the no-match result should happen only after every item has been checked.

A list comprehension builds a new list from an iterable. The expression at the front describes what to save. Read the for clause first, then the expression. Use a normal loop when several steps make a comprehension hard to explain.


#### Example 1 — Transform every number

*Predict the output, then run the code cell below.*


```python
squares_loop = []
for number in range(1, 5):
    squares_loop.append(number * number)
squares_short = [number * number for number in range(1, 5)]
print(squares_loop)
print(squares_short)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>For each number, calculate its square and save it. The comprehension replaces creating an empty list and appending each result.</div>

*Predict the output, then run the code cell below.*


```python
numbers = [1, 2, 3, 4, 5, 6]
evens_loop = []
for number in numbers:
    if number % 2 == 0:
        evens_loop.append(number)
evens_short = [n for n in numbers if n % 2 == 0]
print(evens_loop)
print(evens_short)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>The final if is a filter. Keep only matching items. The saved expression here is the number itself.</div>


#### Example 3 — Transform only matching values

```python
numbers = [-2, 0, 3, 4]
positive_squares = [n * n for n in numbers if n > 0]
print(positive_squares)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><code>set()</code> creates an empty set; <code>add()</code> inserts an item. Curly braces without a key-value colon create a set comprehension. Duplicates collapse into one item. Sorting is only for predictable display.</div>


#### Example 3 — Flatten rows using two for clauses

*Predict the output, then run the code cell below.*


```python
grid = [[1, 2], [3, 4]]
flat_loop = []
for row in grid:
    for value in row:
        flat_loop.append(value)
flat_short = [value for row in grid for value in row]
print(flat_loop)
print(flat_short)
```

First choose a row, then a value in that row. Prefer the full loop when nesting becomes difficult to follow.</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:var(--jp-content-font-color1,#1E293B);"><h2 style="font-size:18px;font-weight:700;line-height:1.35;margin:0;color:inherit;">25. Understand iterators and generators</h2></div>

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br><b>Learner goal.</b> Distinguish a reusable collection from a consumed stream.</div>

An iterator remembers where it is while providing items one at a time. `iter()` gets an iterator; `next()` asks for its next item. A generator expression creates an iterator that computes results as they are requested. This is called lazy evaluation.


#### Example 1 — Request items from an iterator

*Predict the output, then run the code cell below.*


```python
items = iter([10, 20])
print(next(items))
print(next(items))
print(next(items, "Finished"))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>Each next call advances the same iterator. The optional default avoids <code>StopIteration</code> when it is exhausted. A for loop handles exhaustion automatically.</div>


#### Example 2 — A generator is consumed once

*Predict the output, then run the code cell below.*


```python
squares = (n * n for n in range(1, 5))
print(sum(squares))
print(list(squares))
squares = (n * n for n in range(1, 5))
print(list(squares))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br>Parentheses create a generator expression. sum asks for 1, 4, 9, and 16 and adds them. No values remain afterward. Create a fresh generator to produce them again.</div>


#### Example 3 — Stop consuming after a condition is met

*Predict the output, then run the code cell below.*


```python
numbers = iter([2, 4, 7, 8])
has_odd = any(n % 2 != 0 for n in numbers)
print(has_odd)
print(list(numbers))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b><br><code>any()</code> stops when it receives its first <code>True</code> result. It consumes 2, 4, and 7, then stops, leaving 8 in the original iterator.</div>

A list comprehension stores all its results and can be visited again. A generator avoids storing the whole result list; other input data and each produced value still use memory. Use a list when you need indexing or repeated access, and a generator when one-pass processing fits the task.

#### Exercise 6 Find the first even number

*Predict the output, then run the code cell below.*


```python
numbers = [3, 7, 8, 10]
first_even = None
for number in numbers:
    if number % 2 == 0:
        first_even = number
        break
if first_even is None:
    print("No even number")
else:
    print(first_even)
```
