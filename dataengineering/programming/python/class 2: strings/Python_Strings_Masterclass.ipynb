<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Strings as Objects</span></div>

#### 1. What is a String?

**What it is** — An ordered, immutable sequence of characters. Create it with single, double, or triple quotes; an empty string is just `""`.

**Why it matters** — Text is everywhere &mdash; names, files, messages &mdash; so strings are the most-used type in Python.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Triple quotes preserve line breaks exactly, making them ideal for multi-line text and docstrings.
</div>

```python
single = 'hello'
#h   e   l   l   o
#0   1   2   3   4
double = "world"
triple = """line one
line two"""
empty  = ""

print(single, double)
print(triple)
print("empty length:", len(empty))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;All quote styles create the same <code>str</code> type &mdash; choose for readability.
</div>

#### 2. Strings as Sequences &amp; Immutability

**What it is** — A string is a sequence you can measure with `len()` and index into, but you cannot change in place.

**Why it matters** — Immutability is the single fact that explains why every string method returns a *new* string.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
There is no separate <i>character</i> type in Python &mdash; a single character is just a length-1 string.
</div>

```python
s = "Python"

print("length:", len(s))
print("first :", s[0], "| type:", type(s[0]))
print("second :", s[1], "| type:", type(s[0]))

try:
    s[0] = "J"
except TypeError as e:
    print("Immutable ->", e)

print(s)
s="Jython"
print(s)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;A string is a measurable sequence of length-1 strings, fixed once created.
</div>

#### 3. Identity &amp; Type

**What it is** — `type()` reports the class, `isinstance()` tests membership, and `id()` returns the object's identity.

**Why it matters** — These let you verify and branch on what you are actually holding before operating on it.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Prefer <code>isinstance(x, str)</code> over <code>type(x) == str</code> &mdash; it respects subclasses and reads better.
</div>

```python
s = "data"

print(type(s))
print(isinstance(s, str))
print(isinstance(s, int))
print("has identity:", id(s) != 0)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>type</code>/<code>isinstance</code> tell you <i>what</i>; <code>id</code> tells you <i>which</i> object.
</div>

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION II</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Literals &amp; Escape Sequences</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Every way to write a string literal, and the special characters inside them.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">B. Literal Forms</span></div>

#### 1. Raw Strings &amp; Prefixes

**What it is** — Prefixes change how a literal is read: `r` disables escapes, `f` interpolates, `b` makes bytes; they can combine as `rf`.

**Why it matters** — The right prefix prevents whole categories of bugs &mdash; especially in file paths and regex.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Use raw strings <code>r"..."</code> for Windows paths and regex patterns so backslashes mean backslashes.
</div>

```python
name  = "Arun\n"
print(name)
print(name)

path  = r"C:\Users\name\data"   # raw: backslashes literal
fmsg  = f"Hi {name}"               # f: interpolated
both  = rf"{name}\n stays raw"    # rf: combined
print(path)
print(fmsg)
print(both)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>r</code> = literal backslashes, <code>f</code> = embed values, and they stack.
</div>

#### 2. Implicit Concatenation

**What it is** — Adjacent string literals are joined automatically at compile time &mdash; no `+` needed.

**Why it matters** — It lets you split long literals across lines cleanly without runtime cost.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
This works only for <b>literals</b>, not variables &mdash; <code>a b</code> with variables is a syntax error.
</div>

```python
msg = ("Strings placed "
       "side by side "
       "auto-join.")

print(msg)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Bare adjacent literals merge automatically &mdash; great for wrapping long text.
</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">C. Escape Sequences</span></div>

#### 1. Common Escapes

**What it is** — Backslash sequences encode characters you can't type directly: `\n` newline, `\t` tab, `\\` backslash, `\'` `\"` quotes.

**Why it matters** — They control layout and let quotes live inside quotes.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Use <code>\t</code> to align columns and <code>\n</code> to break lines without separate print calls.
</div>

```python
print("Name\tRole")
print("Arun\tEngineer")
print("Quote: \"hi\"")
print("Back\\slash")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;One backslash + a letter = one special character.
</div>

#### 2. Unicode &amp; Hex Escapes

**What it is** — `\u`, `\U`, and `\x` insert characters by their code point or hex value.

**Why it matters** — They let you embed any Unicode character in plain ASCII source.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
Raw strings disable these too &mdash; <code>r"\u0041"</code> prints the literal text, not <code>A</code>.
</div>

```python
print("\u0041\u0042\u0043")   # ABC by code point
print("\x48\x49")             # HI by hex
print("rupee: \u20B9")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>\u</code>/<code>\x</code> address characters by number when you can't type them.
</div>

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION III</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Accessing &amp; Slicing</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Reaching individual characters and extracting ranges by position.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">D. Indexing</span></div>

#### 1. Positive &amp; Negative Indexing

**What it is** — Index `0` is the first character; `-1` is the last. Going past the ends raises `IndexError`.

**Why it matters** — It is the foundation for every position-based text operation.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Negative indices avoid <code>len(s)-1</code> arithmetic &mdash; <code>s[-1]</code> is always the last character.
</div>

```python
s = "PYTHON"

print(s[0], s[-1])      # first, last
print(s[2], s[-2])      # middle access

try:
    print(s[99])
except IndexError as e:
    print("IndexError ->", e)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Front counts from <b>0</b>, back from <b>-1</b>; out-of-range raises <code>IndexError</code>.
</div>

#### 2. Iterating &amp; <code>enumerate()</code>

**What it is** — Looping over a string yields one character at a time; `enumerate()` pairs each with its index.

**Why it matters** — Position-aware iteration powers parsing, counting, and analysis.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Reach for <code>enumerate()</code> instead of <code>range(len(s))</code> &mdash; it is cleaner and less error-prone.
</div>

```python
for i, ch in enumerate("abc"):
    print(i, ch)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>enumerate()</code> gives index <i>and</i> character in one clean loop.
</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">E. Slicing</span></div>

#### 1. <code>[start:stop:step]</code>

**What it is** — Extract a substring. `start` is inclusive, `stop` is exclusive, `step` sets the stride; each can be omitted.

**Why it matters** — Slicing replaces manual loops for prefixes, suffixes, and ranges.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
<code>stop</code> is excluded: <code>s[1:4]</code> returns indices 1, 2, 3 &mdash; not 4.
</div>

```python
s = "Bengaluru"

print(s[0:3])     # prefix
print(s[3:])      # suffix
print(s[::2])     # every 2nd
print(s[:])       # full copy
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Omit a bound for &ldquo;all the way&rdquo;; <code>s[:]</code> copies the whole string.
</div>

#### 2. Reverse &amp; Alternate Extraction

**What it is** — A negative step walks backwards; combining step values pulls out alternating characters.

**Why it matters** — These one-liners cover reversal, sampling, and palindrome checks.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Memorise <code>s[::-1]</code> as the canonical reverse &mdash; every Python developer reads it instantly.
</div>

```python
s = "abcdef"

print(s[::-1])      # reversed
print(s[::2])       # even positions
print(s[1::2])      # odd positions
w = "level"
print(w[::-1])
print("palindrome:", w == w[::-1])
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>step</code> controls direction and stride in a single expression.
</div>

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION IV</div><div style="font-size:22px;font-weight:800;margin-top:2px;">String Operations</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Combining, repeating, testing membership, and comparing strings.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">F. Combining &amp; Testing</span></div>

#### 1. Concatenation &amp; Repetition

**What it is** — `+` joins strings; `*` repeats them an integer number of times.

**Why it matters** — Quick building blocks for labels, separators, and simple layouts.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
You cannot <code>+</code> a string with a number &mdash; convert first with <code>str()</code>.
</div>

```python
print("data" + "-" + "driven")
print("=" * 20)            # multiplier
print("ab" * 3)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>+</code> concatenates; <code>*</code> repeats &mdash; both build new strings.
</div>

#### 2. Membership &amp; Comparison

**What it is** — `in`/`not in` test for substrings; comparison operators order strings lexicographically by code point.

**Why it matters** — Presence checks and ordering drive search, validation, and sorting.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
<code>==</code> compares <i>value</i>; <code>is</code> compares <i>identity</i> &mdash; use <code>==</code> for text equality.
</div>

```python
print("gal" in "Bengaluru")
print("gal" not in "Bengaluru")
print("1" < "2")        # a < b
print("3" > "1")           # uppercase sorts first
print("hi" == "hi")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Use <code>in</code> for presence and <code>==</code> for equality; ordering is by code point.
</div>

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION V</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Core Methods &mdash; Case, Search, Validate</div><div style="font-size:14px;color:#64748B;margin-top:6px;">The everyday method families for changing case, locating text, and checking content.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">G. Case Conversion</span></div>

#### 1. The Case Family

**What it is** — `upper`, `lower`, `capitalize`, `title`, `swapcase`, and `casefold` each return a recased copy.

**Why it matters** — Normalising case makes comparison and display consistent.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Use <code>casefold()</code> (not <code>lower()</code>) for robust case-insensitive matching across languages.
</div>

```python
s = "data SCIENCE"

print(s.upper())
print(s.lower())
print(s.capitalize())
print(s.title())
print(s.swapcase())
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Case methods never mutate &mdash; they hand back a new string.
</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">H. Searching</span></div>

#### 1. Locating Substrings

**What it is** — `find`/`rfind` return an index or `-1`; `index`/`rindex` raise on a miss; `count` tallies occurrences.

**Why it matters** — Finding positions underpins parsing and extraction.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
Prefer <code>find()</code> when a miss is expected &mdash; it returns <code>-1</code> instead of raising.
</div>

```python
s = "go to the goal"

print(s.find("go"))      # first
print(s.rfind("go"))     # last
print(s.count("go"))     # how many
print(s.find("xyz"))     # -1 = absent
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>find</code> signals absence with <code>-1</code>; <code>index</code> raises &mdash; pick by intent.
</div>

#### 2. Edge Matching with Position Args

**What it is** — `startswith`/`endswith` test boundaries and accept start/stop positions to check a slice.

**Why it matters** — Ideal for validating prefixes, suffixes, and file extensions.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Pass a tuple &mdash; <code>name.endswith((".jpg", ".png"))</code> &mdash; to check several options at once.
</div>

```python
f = "report_2024.csv"

print(f.startswith("report"))
print(f.endswith((".csv", ".tsv")))   # tuple of options
print(f.startswith("2024"))         # check from index 7
print(f.startswith("2024", 7))         # check from index 7
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>startswith</code>/<code>endswith</code> accept tuples and positions for flexible checks.
</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">I. Validation</span></div>

#### 1. The <code>is*</code> Family

**What it is** — Methods like `isalpha`, `isdigit`, `isalnum`, `isspace`, `isidentifier` report what a string contains.

**Why it matters** — They replace fragile manual character checks during input validation.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
All <code>is*</code> methods return <code>False</code> for the empty string &mdash; guard for it explicitly.
</div>

```python
print("abc".isalpha())
print("123".isdigit())
print("a1".isalnum())
print("   ".isspace())
print("".isalpha())          # empty -> False
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>is*</code> checks describe content &mdash; always handle the empty-string case.
</div>

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION VI</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Transforming Text</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Cleaning, replacing, splitting, joining, and aligning strings.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">J. Cleaning &amp; Replacing</span></div>

#### 1. Strip, Replace, Remove

**What it is** — `strip` family trims edges, `replace` swaps substrings, `removeprefix`/`removesuffix` peel exact ends.

**Why it matters** — Real input is messy &mdash; trailing spaces and stray markers break logic.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Always <code>strip()</code> user input before validating or comparing it.
</div>

```python
raw = "  user@mail.com  "
print(raw.strip())
print("a-b-c".replace("-", "_"))
print("test_file.txt".removesuffix(".txt"))
print("Mr. Arun".removeprefix("Mr. "))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Trim edges with <code>strip</code>, swap throughout with <code>replace</code>, peel exact ends with <code>remove*</code>.
</div>

#### 2. <code>translate()</code> &amp; <code>maketrans()</code>

**What it is** — `maketrans` builds a character map; `translate` applies it in one fast pass.

**Why it matters** — The fastest way to remap or delete many characters at once.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Great for bulk character substitution &mdash; far faster than chaining many <code>replace()</code> calls.
</div>

```python
table = str.maketrans("aeiou", "AEIOU")
print("education".translate(table))

drop = str.maketrans("", "", "0123456789")
print("abc123def".translate(drop))   # remove digits
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Build a map once with <code>maketrans</code>, apply it in bulk with <code>translate</code>.
</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">K. Splitting &amp; Joining</span></div>

#### 1. Split, Partition, Join

**What it is** — `split`/`splitlines` break text into a list; `partition` splits once into three parts; `join` reassembles.

**Why it matters** — Parsing structured text and rebuilding output are daily tasks.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Build large strings with <code>"".join(parts)</code> &mdash; it is far faster than repeated <code>+</code>.
</div>

```python
print("a,b,c".split(","))
print("line1\nline2".splitlines())
print("key=value".partition("="))
print("-".join(["2024", "01", "31"]))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>split</code>/<code>join</code> are inverses; <code>partition</code> splits exactly once.

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">M. Formatting Approaches</span></div>

#### 1. From <code>%</code> to <code>.format()</code> to f-strings

**What it is** — Python has three styles; f-strings (prefix `f`) are the modern default &mdash; readable and fast.

**Why it matters** — Output quality separates a rough script from a polished tool.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Default to f-strings; reach for <code>%</code> or <code>.format()</code> only when working with legacy code.

```python
name, n = "Arun", 3
print("%s has %d items" % (name, n))      # old
print("{} has {} items".format(name, n))  # format()
print(f"{name} has {n} items")            # f-string
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Three styles, one winner: f-strings for new code.

**What it is** — Format specs after `:` control precision, thousands separators, width, alignment, and notation.

**Why it matters** — One spec replaces manual rounding and spacing.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
<code>{x:,.2f}</code> = thousands separator + 2 decimals &mdash; perfect for currency.

```python
revenue = 1500000.5
print(f"${revenue:,.2f}")          # currency
print(f"{0.875:.1%}")              # percent
print(f"{12345:.2e}")              # scientific
print(f"|{'hi':^8}|")              # centred width 8
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Everything after <code>:</code> is a formatting instruction the value flows into.

#### 3. f-string Expressions &amp; Debug <code>=</code>

**What it is** — f-strings evaluate any expression inside `{}`; appending `=` prints the expression *and* its value.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>

```python
x, y = 7, 3
print(f"{x * y = }")           # debug form
print(f"{x > y = }")
print(f"sum is {x + y}")
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;f-strings run expressions; add <code>=</code> for instant labelled debugging.

```python
print(ord("A"), ord("a"))
print(chr(65), chr(8377))      # A and rupee sign
print("".join(chr(c) for c in range(72, 75)))
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">O. Bytes</span></div>

#### 1. <code>str</code> vs <code>bytes</code>: encode &amp; decode

**What it is** — `str` is human text; `bytes` is raw data. `encode()` turns text into bytes, `decode()` reverses it.

**Why it matters** — Files and networks move bytes, so this boundary is unavoidable.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>

```python
text  = "caf\u00e9"
data  = text.encode("utf-8")
print(data)                     # bytes
print(len(text), len(data))     # chars vs bytes
print(data.decode("utf-8"))     # back to text
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">P. Memory Model</span></div>

#### 1. Interning &amp; Identity

**What it is** — Python reuses (interns) small string-like literals, so identical literals can share one object.

**Why it matters** — It explains surprising `is` results and saves memory for repeated text.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
Compare text with <code>==</code>, never <code>is</code> &mdash; interning is an optimization you must not rely on.
</div>

```python
a = "python"
b = "python"
print(a is b)               # often True (interned)
c = "".join(["py", "thon"])
print(c == a, c is a)       # equal, but maybe not same object
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Interning may share objects; trust <code>==</code> for equality, not <code>is</code>.

#### 2. Object Size with <code>sys.getsizeof()</code>

**What it is** — `sys.getsizeof()` reports the bytes a string object occupies.

**Why it matters** — Useful when profiling memory for large-text workloads.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Longer strings cost more memory; measure before optimising rather than guessing.

```python
import sys
print(sys.getsizeof(""))
print(sys.getsizeof("a"))
print(sys.getsizeof("a" * 100))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Measure real memory cost with <code>sys.getsizeof()</code> instead of assuming.

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">Q. Performance</span></div>

#### 1. Concatenation: <code>+=</code> vs <code>join()</code>

**What it is** — Repeated `+=` rebuilds the whole string each time (O(n²)); collecting parts and `join`-ing is O(n).

**Why it matters** — On large text the difference is the gap between instant and sluggish.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Accumulate pieces in a list and <code>"".join()</code> once at the end.
</div>

```python
parts = [str(i) for i in range(5)]

slow = ""
for p in parts:
    slow += p           # rebuilds each loop
fast = "".join(parts)   # single pass

print(slow, fast, slow == fast)
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Build big strings with one <code>join()</code>, not a loop of <code>+=</code>.
</div>

#### 1. Case-Insensitive Comparison &amp; Sorting

**What it is** — Default ordering is by code point (uppercase before lowercase); normalise case or pass a `key` to sort fairly.

**Why it matters** — Predictable ordering is essential for search, dedup, and display.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">V. Hashing</span></div>

#### 1. <code>hash()</code> and Why Strings Qualify

**What it is** — Because strings are immutable, they have a stable `hash()` &mdash; the requirement for dict keys and set members.

**Why it matters** — It is the reason strings are the default lookup key across Python.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
Mutable objects like lists are unhashable &mdash; only immutable values can be keys or set members.
</div>

```python
print(type(hash("python")))      # int hash

config = {"host": "local", "port": 8080}
print(config["host"])            # string key

tags = {"a", "b", "a"}
print(tags)                       # set dedupes
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Immutability &rarr; stable hash &rarr; valid dict keys and set members.

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">W. Handling Input</span></div>

#### 1. Clean, Validate, Convert

**What it is** — `input()` always returns a string; clean it with `strip`, validate with `is*`, then convert with `int`/`float`.

**Why it matters** — Skipping these steps is the top cause of crashes on real input.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Treat all input as untrusted text: strip &rarr; validate &rarr; convert, with a fallback for bad data.
</div>

```python
def to_age(raw):
    raw = raw.strip()
    if raw.isdigit():
        return int(raw)
    return None

print(to_age("  42 "))     # valid
print(to_age("abc"))       # invalid -> None
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Input is always a string &mdash; clean and validate before converting.
</div>

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION XIV</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Regular Expressions</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Pattern-based search and extraction with the <code>re</code> module.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">X. The <code>re</code> Module</span></div>

#### 1. Search, Findall, Sub

**What it is** — `search` finds the first match, `findall` returns all matches, `sub` replaces matched text.

**Why it matters** — Regex handles flexible patterns that plain methods cannot express.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Write patterns as raw strings (<code>r"..."</code>) so backslashes pass through to the engine intact.
</div>

```python
import re
s = "Call 080-111 or 044-222 today"

print(re.search(r"\d{3}", s).group())   # first match
print(re.findall(r"\d{3}-\d{3}", s))    # all matches
print(re.sub(r"\d", "#", s))             # mask digits
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>search</code> = first, <code>findall</code> = all, <code>sub</code> = replace.
</div>

#### 2. Classes, Quantifiers &amp; Named Groups

**What it is** — Character classes (`\d`, `\w`), quantifiers (`+`, `*`, `{n}`), and `(?P<name>...)` groups capture structure.

**Why it matters** — They turn raw matching into structured extraction.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Named groups make matches self-documenting &mdash; access fields by <code>m.group("name")</code> instead of a number.
</div>

```python
import re
pattern = r"(?P<user>\w+)@(?P<domain>\w+\.\w+)"
m = re.search(pattern, "contact arun@example.com now")
print(m.group("user"))
print(m.group("domain"))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Classes + quantifiers describe shape; named groups extract labelled fields.
</div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">Y. Applied Mini-Projects</span></div>

#### 1. Word &amp; Character Frequency

**What it is** — Combine `split`, a loop or `Counter`, and sorting to summarise text.

**Why it matters** — Frequency analysis is the core of search, analytics, and NLP basics.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
<code>collections.Counter</code> turns frequency counting into a single, readable call.
</div>

```python
from collections import Counter
text = "data is data and data wins"

words = text.split()
print("word count:", len(words))
print("frequencies:", Counter(words))
print("top word:", Counter(words).most_common(1))
```

#### 2. Password Strength Checker

**What it is** — Score a string against rules using `any`, `is*` checks, and length.

**Why it matters** — A compact example of validation logic users meet every day.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Express each rule as a boolean, then combine &mdash; easy to read, test, and extend.
</div>

```python
def strength(pw):
    checks = {
        "length>=8": len(pw) >= 8,
        "has_digit": any(c.isdigit() for c in pw),
        "has_upper": any(c.isupper() for c in pw),
    }
    score = sum(checks.values())
    return score, checks

print(strength("Ab1"))
print(strength("Secure123"))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Model rules as booleans and sum them &mdash; clear, testable validation.
</div>

#### 1. Reverse, Palindrome, Anagram

**What it is** — Three staples: reverse with slicing, palindrome by self-compare, anagram by sorting characters.

**Why it matters** — These appear in nearly every entry-level coding interview.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
Sorting both strings is the simplest anagram test; counting characters is the O(n) upgrade.
</div>

```python
def is_palindrome(s):
    s = s.lower()
    return s == s[::-1]

def is_anagram(a, b):
    return sorted(a.lower()) == sorted(b.lower())

print("reverse:", "python"[::-1])
print("palindrome:", is_palindrome("Level"))
print("anagram:", is_anagram("listen", "silent"))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Slicing, self-compare, and sorted-compare solve the three classic string problems.
</div>

#### 2. Most Frequent Character &amp; Compression

**What it is** — Find the top character with `Counter`; run-length compress repeated runs.

**Why it matters** — Common follow-up questions that test loops and dictionaries.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
Run-length compression only saves space when runs are long &mdash; return the original if it does not.
</div>

```python
from collections import Counter

def most_frequent(s):
    return Counter(s.replace(" ", "")).most_common(1)[0]

def compress(s):
    out, i = [], 0
    while i < len(s):
        j = i
        while j < len(s) and s[j] == s[i]:
            j += 1
        out.append(s[i] + str(j - i))
        i = j
    result = "".join(out)
    return result if len(result) < len(s) else s

print(most_frequent("mississippi"))
print(compress("aaabbc"))
```

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>Counter</code> finds peaks; run-length encoding shrinks long repeats.
</div>

#### 1. The Frequent Traps

**What it is** — Out-of-range indexing, expecting mutation, `is` for equality, and forgetting `input()` returns text.

**Why it matters** — Knowing the usual failure points lets you spot bugs before they ship.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
These four mistakes account for a large share of beginner string bugs &mdash; learn to recognise each on sight.

```python
# 1. expecting mutation (it never happens)
s = "abc"; s.upper(); print("unchanged:", s)
# 2. is vs == on built strings
x = "ab" * 1; y = "abab"[:2]
print("== correct:", x == y, "| is risky:", x is y)
# 3. input() always returns str
print("type from input:", type("123"))
```
