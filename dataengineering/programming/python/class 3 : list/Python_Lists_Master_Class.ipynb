## One name, many related values

Imagine writing a shopping list. Giving every item a separate variable is awkward:

```python
item_1 = "milk"
item_2 = "bread"
item_3 = "eggs"
```

A Python **list** keeps the related values together, in order, under one name.

<div style="background:#EFF6FF;border-left:5px solid #2563EB;padding:11px 14px;border-radius:7px;color:#1E293B;"><b>Plain-language idea</b><br>A list is like a numbered row of boxes kept under one name. Each box holds one item. Python remembers the order of the boxes, and the row can be changed, expanded, or shortened.</div>

````python
shopping = ["milk", "bread", "eggs"]
print(shopping)
print(type(shopping))
````

## Definition of a Python list

> **Formal definition:** A Python list is a built-in, **ordered**, **mutable sequence** that can store zero or more Python objects. Lists allow duplicate values, may contain objects of different types, and can grow or shrink while a program is running.

In simpler words: a list is one Python value that keeps several items together in a particular order and allows us to change that collection.

Now let us unpack every important word in the definition.

### 1. A list is a collection or sequence

A **collection** groups several values into one value. A **sequence** is a collection whose items have positions: first, second, third, and so on.

The list below is one value named `shopping`, and it contains three items.

````python
shopping = ["milk", "bread", "eggs"]
print("The whole list:", shopping)
print("Number of items:", len(shopping))
````

### 2. A list is ordered

**Ordered** means Python preserves the position of every item. If `"wake up"` is placed first and `"eat breakfast"` is placed third, those positions remain meaningful.

Ordered does **not** mean automatically sorted. Python does not rearrange a list alphabetically or numerically unless we ask it to.

````python
morning_steps = ["wake up", "brush teeth", "eat breakfast"]

print(morning_steps)
print("First step:", morning_steps[0])
print("Last step:", morning_steps[-1])
````

### 3. A list is mutable

**Mutable** means the existing list can be changed after it has been created. We can replace an item, add a new item, or remove an item.

The variable still refers to the same list object; only its contents change.

````python
tasks = ["draft", "review"]
print("Before:", tasks)

tasks[0] = "write"
print("After: ", tasks)
````

### 4. A list allows duplicate values

The same value may appear more than once. Each occurrence occupies its own position.

````python
votes = ["yes", "no", "yes", "yes"]
print(votes)
print("Number of yes votes:", votes.count("yes"))
````

### 5. A list may contain different types of objects

Python allows text, numbers, Boolean values, and even other lists inside one list. This is sometimes useful, but real programs are usually clearer when all items serve the same purpose.

````python
person_summary = ["Ravi", 28, True,[1,2]]
temperatures = [31.5, 32.0, 30.8]
empty_list = []

print(person_summary)
print(temperatures)
print(empty_list)
````

### 6. A list has a dynamic size

**Dynamic size** means the list can grow when we add items and shrink when we remove items. Its length is not fixed when it is created.

````python
guests = ["Asha"]
print("Starting list:", guests)

guests.append("Ben")
print("After adding:", guests)

guests.remove("Asha")
print("After removing:", guests)
````

## Every position has an index

People count the first item as **1**. Python counts the first item as **0**.

| Item | tea | coffee | juice | water |
|---|---:|---:|---:|---:|
| Index from front | 0 | 1 | 2 | 3 |
| Index from back | -4 | -3 | -2 | -1 |

Use `list_name[index]` to read one item.

````python
drinks = ["tea", "coffee", "juice", "water"]

print("First:", drinks[0])
print("Second:", drinks[1])
print("Last:", drinks[-1])
print("Second from last:", drinks[-2])
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — An index outside the list raises `IndexError`. A four-item list has front indexes 0, 1, 2, and 3—not 4.</div>

````python
drinks = ["tea", "coffee", "juice", "water"]

try:
    print(drinks[4])
except IndexError:
    print("There is no item at index 4.")
````

## Three common ways to grow a list

- `append(value)` adds **one item at the end**.
- `insert(index, value)` adds **one item at a chosen position**.
- `extend(other_list)` adds **every item from another collection**.

````python
guests = ["Asha", "Ben"]

guests.append("Chen")          # one item at the end
print(guests)
guests.insert(1, "Divya")     # one item at index 1
guests.extend(["Eli", "Fatima"])  # many items

print(guests)
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — `append(["Eli", "Fatima"])` adds the whole inner list as one item. `extend(["Eli", "Fatima"])` adds two separate items.</div>

````python
a = [1, 2]
a.append([3, 4])
print("append:", a)

b = [1, 2]
b.extend([3, 4])
print("extend:", b)
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — Append one item, insert one item at a position, and extend with many items.</div>

## Choose removal by what you know

- Know the **value**? Use `remove(value)`.
- Know the **position**, and may need the removed item? Use `pop(index)`.
- Want to delete by position without keeping the value? Use `del`.
- Want an empty list? Use `clear()`.

````python
queue = ["Asha", "Ben", "Chen", "Divya", "Ben"]

queue.remove("Ben")       # remove by value
print("remove:", queue)
served = queue.pop(0)  # remove by index and return the value
print(queue)
del queue[-1]             # delete by index

print("Served:", served)
print("Waiting:", queue)
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — `remove()` raises `ValueError` when the value is absent, and `pop()` raises `IndexError` when the index is invalid. Check first when absence is normal.</div>

````python
cart = ["rice", "soap"]
item_to_remove = "oil"

if item_to_remove in cart:
    cart.remove(item_to_remove)
else:
    print(item_to_remove, "is not in the cart")

print(cart)
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — Remove by value with `remove`; remove by position and receive the value with `pop`.</div>

````python
tasks = ["open file", "clean data", "save report"]

for number, task in enumerate(tasks, start=1):
    print(number, task)
````

### Build a result step by step

Start with an empty list, compute one result at a time, and append it.

````python
prices = [50, 120, 80]
discounted_prices = []

for price in prices:
    discounted_prices.append(price * 0.90)

print(discounted_prices)
````

## The pattern `[start:stop:step]`

A slice returns a new list.

- `start` is included.
- `stop` is excluded.
- `step` says how far to jump.
- Any part may be omitted.

````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print("Weekdays:", days[0:5])
print("Weekend: ", days[5:])
print("First 3: ", days[:3])
print("Every 2nd:", days[::2])
print("Reversed:", days[::-1])
````

<div style="background:#EFF6FF;border-left:5px solid #2563EB;padding:11px 14px;border-radius:7px;color:#1E293B;"><b>Plain-language idea</b><br>Think of a slice as cutting *between* boxes. `days[1:4]` starts at box 1 and stops at the boundary before box 4.</div>

````python
numbers = [10, 20, 30, 40, 50]
middle = numbers[1:4]
print(middle)
print("Original stays unchanged:", numbers)
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — A slice reads a range. The start is included, the stop is excluded, and omitted values mean “from the edge.”</div>

## `sorted()` versus `.sort()`

- `sorted(items)` returns a **new sorted list**; the original is unchanged.
- `items.sort()` changes the **existing list** and returns `None`.

For beginners, `sorted()` is often safer because it keeps the original data.

````python
scores = [72, 91, 65, 88]

low_to_high = sorted(scores)
high_to_low = sorted(scores, reverse=True)

print("Original:", scores)
print("Ascending:", low_to_high)
print("Descending:", high_to_low)
````

````python
scores = [72, 91, 65, 88]

print("Count:", len(scores))
print("Total:", sum(scores))
print("Lowest:", min(scores))
print("Highest:", max(scores))
print("Average:", sum(scores) / len(scores))
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — `min()`, `max()`, and average calculations need at least one item. Check `if scores:` before using them when the list may be empty.</div>

````python
names = ["Zoya", "arun", "Meera"]
print(sorted(names, key=str.casefold))  # ignores letter case while sorting
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — Use `sorted()` when you want a new order without changing the original. Built-ins such as len, sum, min, and max make common summaries readable.</div>

## Assignment does not copy a list

When you write `backup = original`, both names point to the same list. Changing through either name changes that one shared object.

````python
original = ["draft", "review"]
backup = original

backup.append("publish")
print("original:", original)
print("backup:  ", backup)
print("Same object?", original is backup)
````

Use `.copy()` when you need a separate top-level list.

````python
original = ["draft", "review"]
backup = original.copy()

backup.append("publish")
print("original:", original)
print("backup:  ", backup)
print("Same object?", original is backup)
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — `.copy()` is shallow: inner lists are still shared. Deep copying is an advanced tool and is only needed when nested mutable data must be fully independent.</div>

````python
import copy

matrix = [[1, 2], [3, 4]]
deep_backup = copy.deepcopy(matrix)
deep_backup[0][0] = 99

print("matrix:", matrix)
print("backup:", deep_backup)
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — `=` creates another name for the same list. `.copy()` creates a separate outer list; `deepcopy()` also separates nested mutable items.</div>

## A list can be an item inside another list

This is useful for rows and columns. Use one index to choose a row and a second index to choose an item in that row.

````python
seats = [
    ["A1", "A2", "A3"],
    ["B1", "B2", "B3"],
]

print("First row:", seats[1])
print("Seat in row 2, column 3:", seats[1][2])
````

````python
for row in seats:
    for seat in row:
        print(seat, end=" ")
    print()
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — Avoid `grid = [[0] * 3] * 2`. It repeats references to the same inner list, so changing one row changes the others.</div>

````python
bad_grid = [[0] * 3] * 2
bad_grid[0][0] = 9
print("Shared rows:", bad_grid)

good_grid = [[0] * 3 for _ in range(2)]
good_grid[0][0] = 9
print("Independent rows:", good_grid)
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — Use `[row][column]` for a nested list, and build mutable rows one at a time with a comprehension.</div>

## First understand the loop

This loop creates a list of squares:

````python
numbers = [1, 2, 3, 4, 5]
squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
````

A **list comprehension** expresses the same idea in one line:

```python
[result for item in source]
```

Read it from the middle: “for each `number` in `numbers`, produce `number ** 2`.”

````python
numbers = [1, 2, 3, 4, 5]
squares = [number ** 2 for number in numbers]
print(squares)
````

Add an `if` at the end to keep only matching items:

```python
[item for item in source if condition]
```

````python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [number for number in numbers if number % 2 == 0]
print(even_numbers)
````

## The `list()` constructor

Square brackets are clearest when you already know the items. `list(iterable)` collects values produced by another iterable.

- `list("cat")` collects the characters from a string.
- `list(range(1, 6))` collects the numbers produced by a range.
- `list(a_tuple)` converts a tuple to a list.

An **iterable** is simply something Python can visit one item at a time.

````python
print(list("cat"))
print(list(range(1, 6)))
print(list(("red", "green", "blue")))
````

## Combine and repeat

`+` creates a new list by concatenating two lists. `*` repeats the references stored in a list.

````python
first = [1, 2]
second = [3, 4]

combined = first + second
zeros = [0] * 5

print(combined)
print(zeros)
print("first is unchanged:", first)
````

## Mutating methods usually return `None`

Methods that mainly change a list in place—such as `append`, `extend`, `insert`, `remove`, `clear`, `reverse`, and `sort`—do not normally return the changed list. This prevents code from hiding an important mutation.

````python
numbers = [3, 1, 2]
result = numbers.sort()

print("numbers:", numbers)
print("result:", result)
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — Do not write `numbers = numbers.sort()`. After that statement, `numbers` becomes `None`. Use `numbers.sort()` alone or `numbers = sorted(numbers)`.</div>

## A function receives a reference to the same list

If a function mutates its list parameter, the caller sees the change.

````python
def add_status(items):
    items.append("done")

tasks = ["draft", "review"]
add_status(tasks)
print(tasks)
````

Make the contract clear. Either name the function and document that it changes the input, or create and return a new list.

````python
def with_status(items):
    return items + ["done"]

tasks = ["draft", "review"]
updated_tasks = with_status(tasks)
print("original:", tasks)
print("new:     ", updated_tasks)
````

## The mutable default argument trap

Default argument values are created once, when Python defines the function—not each time the function is called. Never use a list literal as a default when each call should receive a fresh list.

````python
# Wrong: one default list is shared across calls.
def collect_wrong(name, names=[]):
    names.append(name)
    return names

print(collect_wrong("Asha"))
print(collect_wrong("Ben"))
````

````python
# Correct: create a new list inside the call.
def collect(name, names=None):
    if names is None:
        names = []
    names.append(name)
    return names

print(collect("Asha"))
print(collect("Ben"))
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — Mutation changes the existing object. Mutating methods usually return `None`, functions can mutate the caller’s list, and `None` is the safe default for an optional list parameter.</div>

## Walk related lists together with `zip()`

`zip()` produces pairs (or larger groups) from matching positions.

````python
names = ["Asha", "Ben", "Chen"]
scores = [82, 91, 76]

for name, score in zip(names, scores):
    print(name, "->", score)
````

<div style="background:#FFF8E1;border-left:5px solid #D59A00;padding:10px 14px;border-radius:7px;color:#463500;"><b>Watch out</b> — `zip()` stops when the shortest iterable ends. If unequal lengths indicate bad data, check the lengths first or use `itertools.zip_longest`.</div>

## Reverse iteration without changing the list

`reversed(items)` produces values from the end to the beginning. It does not mutate the list.

````python
steps = ["start", "work", "finish"]
for step in reversed(steps):
    print(step)

print("Original:", steps)
````

## `any()` and `all()`

- `any(condition for item in items)` asks whether at least one item passes.
- `all(condition for item in items)` asks whether every item passes.

````python
scores = [82, 91, 47, 76]
print("Any distinction?", any(score >= 90 for score in scores))
print("All passed?", all(score >= 50 for score in scores))
````

Unpacking places items into variables. The starred variable collects the remaining values into a list.

## `==` compares values; `is` compares identity

Two separate lists may contain equal values. Use `==` for normal data comparison. Use `is` mainly for singleton objects such as `None`, or while demonstrating shared references.

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)
print("a is b:", a is b)
print("a is c:", a is c)
````

Lists also support lexicographic ordering when corresponding items can be compared: Python compares the first different position. This is useful occasionally, but a `key=` function is clearer for most records.

````python
print([1, 2] < [1, 3])
print([1, 9] < [2, 0])
````

## Why lists are unhashable

A dictionary key or set member needs a stable hash. Because a list is mutable, its value can change, so Python does not allow a list to be used directly as a dictionary key or set member. Use a tuple when the sequence is fixed and its contents are hashable.

````python
try:
    lookup = {[1, 2]: "point"}
except TypeError as error:
    print("List key error:", error)

lookup = {(1, 2): "point"}
print(lookup[(1, 2)])
````

<div style="background:#ECFDF5;border-left:5px solid #16835B;padding:10px 14px;border-radius:7px;color:#1E293B;"><b>Key takeaway</b> — Use `==` to compare contents and `is` to test whether two names reference the same object. Lists are unhashable because they are mutable.</div>

## Lists are dynamic arrays

You do not need the implementation details for everyday code, but the shape explains the common performance rules:

- read or replace by index: **O(1)**;
- append or pop from the end: **O(1) amortized**;
- search with `in`, `index`, or `remove`: **O(n)**;
- insert or pop near the front: **O(n)** because later references must shift.

`O(1)` means the work stays roughly constant as the list grows. `O(n)` means the work may grow with the number of items.

## Stack: a good use of a list

A stack uses last-in, first-out order. `append()` and `pop()` at the end are efficient.

````python
stack = []
stack.append("page 1")
stack.append("page 2")
print("Removed:", stack.pop())
print("Left:", stack)
````

## Queue: prefer `deque`

Repeated `pop(0)` is costly because every remaining item shifts. For a first-in, first-out queue, use `collections.deque`.

````python
from collections import deque

queue = deque(["Asha", "Ben", "Chen"])
print("Served:", queue.popleft())
print("Waiting:", list(queue))
````

## A list of dictionaries for records

When every record has named fields, a dictionary is clearer than remembering that index 0 means name and index 1 means score.

````python
students = [
    {"name": "Asha", "score": 82},
    {"name": "Ben", "score": 91},
    {"name": "Chen", "score": 76},
]

top_first = sorted(students, key=lambda student: student["score"], reverse=True)
for student in top_first:
    print(student["name"], student["score"])
````

## Common mistakes

### 1. Forgetting that indexing starts at zero

For `items = ["a", "b", "c"]`, the last valid front index is `2`, not `3`.

### 2. Confusing `append` and `extend`

Append adds one value. Extend adds each value from another iterable.

### 3. Expecting `.sort()` to return the sorted list

`.sort()` changes the list and returns `None`. Use `sorted(items)` when you want a new list.

### 4. Accidentally sharing a list

`copy = original` shares one list. Use `copy = original.copy()` for a separate outer list.

### 5. Changing a list while looping over it

Build a new filtered list instead.

### 6. Using a list when a dictionary is clearer

If you keep asking “which value belongs to this name?”, a dictionary may communicate the data better.

### 7. Using `is` for value comparison

Use `==` to compare list contents. `is` asks whether two names refer to the exact same object.

### 8. Using a list as a default argument

Use `None` as the default, then create a new list inside the function.

### 9. Assigning a mutating method result

Do not write `items = items.sort()` or `items = items.append(value)`. These methods mutate the list and return `None`.
