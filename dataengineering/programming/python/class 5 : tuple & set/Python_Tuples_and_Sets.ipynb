#### Tuple literals & packing

**What it is** — A tuple is an ordered collection. Commas create the tuple; parentheses mainly group it.

**Why it matters** — Tuples keep related values together and can return several results.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; The comma is the tuple-maker.</div>

**Python Example**

```python
point=(10,20)
packed=1,2,3
print(point)
print(packed)
print(type(point))
print(type(packed))
```

````python
point=(10,20)
packed=1,2,3
print(point)
print(packed)
print(type(point))
print(type(packed))
````

#### The single-element trap

**What it is** — A one-item tuple needs a trailing comma: (5,). Without it, (5) is the integer 5.

**Why it matters** — This common beginner bug changes the type silently.

<div style="background:#F1F5F9;border-left:5px solid #64748B;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>Important Note</b>&nbsp; (5) is the number 5; (5,) is a one-element tuple.</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; For one item, the comma is mandatory.</div>

**Python Example**

```python
not_tuple=(5)
yes_tuple=(5,)
print(type(not_tuple))
print(type(yes_tuple))
print(len(yes_tuple))
```

**Output**

```text
<class 'int'>
<class 'tuple'>
1
```

````python
not_tuple=("s")
yes_tuple=(5,)
print(type(not_tuple))
print(type(yes_tuple))
print(len(yes_tuple))
````

#### The tuple() constructor

**What it is** — tuple() builds a tuple from an iterable such as a list, string, range, or set.

**Why it matters** — It freezes an existing iterable into an immutable sequence.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Use tuple(iterable) when the source already exists.</div>

**Python Example**

```python
print(tuple([1,2,3]))
print(tuple("cat"))
print(tuple(range(3)))
```

**Output**

```text
(1, 2, 3)
('c', 'a', 't')
(0, 1, 2)
```

````python
print(tuple([1,2,3]))
print(tuple("cat"))
print(tuple(range(3)))
````

#### Indexing and negative indexing

**What it is** — Tuple positions start at zero; negative indexes count backward.

**Why it matters** — Indexing retrieves a known position without changing the tuple.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; t[0] is first; t[-1] is last.</div>

**Python Example**

```python
colors=("red","green","blue")
print(colors[0])
print(colors[-1])
```

````python
colors=("red","green","blue")
print(colors[1])
print(colors[-3])
````

#### Slicing tuples

**What it is** — A slice uses start:stop:step; stop is excluded.

**Why it matters** — It creates a new tuple while leaving the original untouched.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; A tuple slice is a new tuple.</div>

**Python Example**

```python
nums=(0,1,2,3,4,5)
print(nums[1:4])
print(nums[::2])
print(nums[::-1])
```


````python
nums=(0,1,2,3,4,5)
print(nums[1:5])
print(nums[::3])
print(nums[::-2])
````

#### Membership, length, and counting

**What it is** — in, len(), count(), and index() inspect tuple contents.

**Why it matters** — You can answer common questions without converting to a list.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Use in for yes/no, count for frequency, index for position.</div>

**Python Example**

```python
scores=(80,95,80,70)
print(95 in scores)
print(len(scores))
print(scores.count(80))
print(scores.index(70))
```

````python
scores=(80,95,80,70)
print(95 in scores)
print(len(scores))
print(scores.count(80))
print(scores.index(70))
````

#### Unpacking

**What it is** — Unpacking assigns tuple items to several variables in one statement.

**Why it matters** — It makes fixed-format data and function results readable.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Variables must match values unless starred unpacking is used.</div>

**Python Example**

```python
point=(10,20)
x,y=point
print(x,y)
first,*middle,last=(1,2,3,4,5)
print(first,middle,last)
```

**Output**

```text
10 20
1 [2, 3, 4] 5
```

````python
point=(10,20)
x,y=point
print(x,y)
first,*middle,last=(1,2,3,4,5)
print(first,middle,last)
````

#### Tuples from functions

**What it is** — A function can return multiple values; Python packages them as a tuple.

**Why it matters** — Related results stay together and callers can unpack them.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Multiple return values are an everyday tuple pattern.</div>

**Python Example**

```python
def min_max(values):
    return min(values),max(values)
low,high=min_max([4,9,2,7])
print(low,high)
```

**Output**

```text
2 9
```

````python
def min_max(values):
    return min(values),max(values)
low,high=min_max([4,9,2,7])
print(low,high)
````

#### Immutability is shallow

**What it is** — A tuple cannot replace its slots, but a nested list can still change internally.

**Why it matters** — This separates immutable structure from mutable objects inside it.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; The tuple is fixed; nested objects keep their own mutability.</div>

**Python Example**

```python
data=("team",["Arun","Bina"])
data[1].append("Carl")
print(data)
try: data[0]="group"
except TypeError as e: print(type(e).__name__)
```

````python
data=("team",["Arun","Bina"])
print(data)
data[0]="team2"
````

````python
data=("team",["Arun","Bina"])
print(type(data[1]))
data[1].append("Carl")
print(data)
````

**What it is** — A tuple can be a dictionary key when all contents are hashable.

**Why it matters** — Coordinates and composite IDs often make good keys.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; A tuple is hashable only when every nested value is hashable.</div>

**Python Example**

```python
locations={(18.52,73.85):"Pune"}
print(locations[(18.52,73.85)])
```

**Output**

```text

```

````python
locations={(18.52,73.85):"Pune"}
print(locations[(18.52,73.85)])
````

#### Set literals and set()

**What it is** — A set stores unique values. Curly braces with values make a set; set() makes an empty set or converts an iterable.

**Why it matters** — Sets remove duplicates and make membership efficient.

<div style="background:#F1F5F9;border-left:5px solid #64748B;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>Important Note</b>&nbsp; Set display order is not a teaching contract.</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; {} is an empty dict; use set() for an empty set.</div>

````python
tags={"python","data","python"}
dict_val = {"key":"value"}
print(tags)
print(type(tags))
print(type(dict_val))
````

````python
print(set([1,2,2,3]))
````

````python
tags={"python","data","python"}

tags.add("data")
tags.add("set")
print(tags)
````

#### Sets are unordered

**What it is** — A set has no indexing or meaningful positional order.

**Why it matters** — If you need positions, use a list or tuple.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Use a set when uniqueness matters more than position.</div>


````python
items={"pen","book","bag"}
print("pen" in items)
print(len(items))
try: print(items[0])
except TypeError as e: print(type(e).__name__)
````

#### Deduplicate a sequence

**What it is** — Converting a list to a set removes repeated values.

**Why it matters** — It is a compact way to clean repeated membership data.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; list(set(values)) removes duplicates but not guaranteed order.</div>


````python
values=["A","B","A","C","B"]
unique=set(values)
print(unique)
print(len(unique))
````

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION V</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Set Algebra</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Combine groups with union, intersection, difference, and symmetric difference.</div></div>

#### Union: combine everything

**What it is** — The union | contains elements from either set.

**Why it matters** — It answers “who is in at least one group?” without duplicates.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Union combines membership.</div>


````python
python={"Arun","Bina","Carl"}
data={"Bina","Carl","Diya"}
print(python|data)
print(sorted(python|data))
````

#### Intersection: keep shared items

**What it is** — The intersection & contains only elements present in both sets.

**Why it matters** — It answers “who belongs to both groups?” directly.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Intersection finds overlap.</div>

````python
python={"Arun","Bina","Carl"}
data={"Bina","Carl","Diya"}
print(sorted(python&data))
````

#### Difference: what one group has

**What it is** — The difference - keeps items in the left set absent from the right.

**Why it matters** — It answers “what is unique to this group?”; direction matters.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; a-b is not b-a.</div>

````python
python={"Arun","Bina","Carl"}
data={"Bina","Carl","Diya"}
print(sorted(python-data))
print(sorted(data-python))
````

#### Symmetric difference

**What it is** — The symmetric difference ^ keeps items in exactly one set.

**Why it matters** — It reveals disagreement between two memberships.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; It means “either group, but not the overlap.”</div>

````python
a={1,2,3}; b={3,4,5}
print(sorted(a^b))
````

#### Subsets, supersets, and disjoint sets

**What it is** — <=, >=, and isdisjoint() test relationships between groups.

**Why it matters** — These validate permissions, categories, and requirements.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; small <= big means every item in small is in big.</div>

**Python Example**

```python
required={"read","write"}
user={"read","write","share"}
print(required<=user)
print(user>=required)
print(required.isdisjoint({"admin"}))
```

**Output**

```text
True
True
True
```

````python
required={"read","write"}
user={"read","write","share"}
print(required<=user)
print(user>=required)
print(required.isdisjoint({"admin"}))
````

#### add() and update()

**What it is** — add(x) inserts one item; update(iterable) inserts many.

**Why it matters** — The method shows whether you add one member or merge a group.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Adding an existing value has no effect.</div>

````python
skills={"Python"}
skills.add("SQL")
skills.update(["Git","Python","Java"])
print(sorted(skills))
````

#### remove(), discard(), pop(), clear()

**What it is** — remove raises if missing; discard does not. pop removes an arbitrary item; clear empties.

**Why it matters** — Choosing the remover makes missing-data behaviour explicit.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Use discard for safe optional removal; never assume pop is first.</div>

````python
s={"a","b","z"}
s.pop()
print(s)
````

````python
s={"a","b","z"}
print(s)
s.discard("z")
print(s)
s.discard("k")
````

````python
s={"a","b","z"}
print(s)
s.remove("z")
print(s)
s.remove("k")
````

````python
s={"a","b","z"}
s.clear()
print(s)
````

#### Stable display order

**What it is** — Set iteration order is not guaranteed; use sorted(s) for stable display.

**Why it matters** — Sorting at the boundary makes reports and tests reproducible.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Keep the set for membership; sort only when presenting.</div>

**Python Example**

```python
cities={"Pune","Delhi","Chennai"}
for city in sorted(cities): print(city)
```

**Output**

```text
Chennai
Delhi
Pune
```

````python
cities={"Pune","Delhi","Chennai"}
for city in sorted(cities): print(city)
````

#### Set comprehensions

**What it is** — A set comprehension builds a unique set with an expression and optional filter.

**Why it matters** — It combines transformation, filtering, and deduplication.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:9px 11px;border-radius:6px;color:#1E293B;"><b>🎯 Key Takeaway</b>&nbsp; Use {expression for item in iterable if condition}.</div>

**Python Example**

```python
words=["cat","car","dog","cup"]
starts={w[0] for w in words if w.startswith("c")}
print(starts)
```

**Output**

```text
{'c'}
```

````python
words=["cat","car","dog","cup"]
starts={w[0] for w in words if w.startswith("c")}
print(starts)
````

### Common Mistakes

- Writing `(5)` when you mean `(5,)`.
- Expecting a set to support indexing or preserve a useful display order.
- Using `remove()` when a missing item should be harmless; choose `discard()`.
- Treating a tuple containing a list as completely immutable.
- Using a set when order or duplicates are meaningful.

### Final Takeaways

Tuples are ordered and structurally immutable. Sets are unique and unordered; use them for membership and group operations. Choose a collection by the behaviour your program needs.
