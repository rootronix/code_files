#### 1. What is a Dictionary?

**What it is** — A dictionary maps <b>keys</b> to <b>values</b> and finds a value by its key: <code>{"name": "Arun", "age": 25}</code>. It is Python’s built-in mapping type and the structure you reach for whenever data has a natural label.

**Why it matters** — Four properties define a dict — mutability, ordering, keyed access and unique hashable keys. Knowing them up front tells you when a dict is the right tool and how it will behave.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
<b>Properties.</b> A dict is <b>mutable</b> (add, change and remove entries in place), <b>ordered</b> (keys keep their insertion order, guaranteed since Python 3.7), <b>keyed</b> rather than positionally indexed (you look items up by key, not by an integer position), and holds <b>unique, hashable keys</b> (each key appears once; values can be anything and may repeat).
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;A dictionary is a <b>mutable</b>, <b>insertion-ordered</b> map of <b>unique, hashable keys</b> to values, with O(1) lookup by key.
</div>

```python
d = {"name": "Arun", "age": 25}
print(d)
d["age"] = 26              # mutable: change a value in place
print(d)                   # ordered: insertion order is kept
d["city"] = "Pune"         # mutable: add a new key
print(d)
print(list(d))             # keys, in insertion order
print(d["name"])           # keyed access, not by position
print(d)
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Dictionary Basics</span></div>

#### 1. Dict Literals & dict()

**What it is** — A dict maps keys to values: <code>{"name": "Arun"}</code>. The <code>dict()</code> constructor accepts keyword pairs too.

**Why it matters** — Dictionaries model labelled data — records, configs, lookups — everywhere in Python.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Literals use <code>{key: value}</code>; <code>dict(key=value)</code> is a clean keyword alternative.
</div>

```python
person = {"name": "Arun", "age": 25}
new = dict(name="Arun", age=25)
print(new)
print(person == new)
```

#### 2. Building from Pairs, zip & fromkeys

**What it is** — <code>dict()</code> accepts a list of <code>(key, value)</code> pairs; <code>zip()</code> pairs two sequences; <code>fromkeys()</code> seeds shared defaults.

**Why it matters** — These patterns turn parallel data into a mapping without a manual loop.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>dict(zip(keys, values))</code> and <code>dict.fromkeys(keys, default)</code> build maps in one line.
</div>

```python
keys = ["a", "b", "c"]
vals = [1, 2, 3]
print(dict(zip(keys, vals)))
```

#### 3. Dict Comprehensions

**What it is** — <code>{k: v for ...}</code> builds a dictionary declaratively from any iterable.

**Why it matters** — It transforms or filters data into a map in a single expression.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Dict comprehensions construct mappings inline — clear and fast.
</div>

```python
squares = {n: n * n for n in range(1, 5)}
print(squares)
evens = {n: n for n in range(6) if n % 2 == 0}
print(evens)
```

#### 4. Nested Dictionaries

**What it is** — Values can themselves be dicts, modelling hierarchical data such as JSON.

**Why it matters** — Most real-world configuration and API data is naturally nested.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Nest dicts to model trees; index level by level to reach deep values.
</div>

```python
users = {
    "u1": {"name": "Arun", "roles": ["admin"]},
    "u2": {"name": "Bina", "roles": ["editor"]},
}
print(users["u1"])
print(users["u1"]["name"])
print(users["u2"]["roles"])
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Reading & Writing</span></div>

#### 1. Indexing [ ] vs get()

**What it is** — <code>d[key]</code> raises <code>KeyError</code> if the key is missing; <code>d.get(key, default)</code> returns a fallback instead.

**Why it matters** — <code>get()</code> lets you read optional data without wrapping every access in try/except.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Use <code>d.get(key, default)</code> whenever a key might be absent; reserve <code>d[key]</code> for keys you know exist.
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>[]</code> demands the key exists; <code>get()</code> returns a default when it doesn’t.
</div>

```python
d = {"a": 1}
print(d["a"])
print(d.get("a"))
print(d.get("z"))
print(d.get("z", 0))
try:
    d["z"]
except KeyError as e:
    print("KeyError:", e)
```

#### 2. Adding & Updating Keys

**What it is** — Assigning to a new key inserts it; assigning to an existing key overwrites its value.

**Why it matters** — This single syntax both creates and edits entries.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>d[key] = value</code> inserts if new, overwrites if the key already exists.
</div>

```python
d = {"a": 1}
d["b"] = 2          # insert
d["a"] = 99         # overwrite
print(d)
```

#### 3. setdefault()

**What it is** — <code>setdefault(key, default)</code> returns the existing value, or inserts and returns the default if the key is missing.

**Why it matters** — It is the clean way to build a dict-of-lists without a membership check.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>setdefault()</code> fetches-or-initialises a key in one atomic step.
</div>

```python
groups = {}
for name, team in [("Arun", "A"), ("Bina", "B"), ("Carl", "A")]:
    groups.setdefault(team, []).append(name)
print(groups)
```

#### 4. Merging: update(), | and {**a, **b}

**What it is** — <code>update()</code> merges another mapping in place; Python 3.9+ adds <code>|</code> (new dict) and <code>|=</code> (in place); <code>{**a, **b}</code> spreads both into a fresh dict.

**Why it matters** — Merging configs and overriding defaults is an everyday task.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
In every form the right-hand mapping wins on conflicting keys — so put overrides last.
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>update()</code>/<code>|=</code> merge in place; <code>|</code> and <code>{**a, **b}</code> return a new merged dict.
</div>

```python
base = {"a": 1, "b": 2}
over = {"b": 20, "c": 3}
print(base | over)        # new dict, right wins
print({**base, **over})   # same result via spread
base.update(over)
print(base)
```

#### 5. Deleting: del, pop(), popitem(), clear()

**What it is** — <code>del d[k]</code> and <code>pop(k)</code> remove by key (pop returns the value); <code>popitem()</code> removes the last pair; <code>clear()</code> empties the dict.

**Why it matters** — Different removers fit different needs — return a value, take any item, or reset.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>pop()</code> returns the removed value; <code>popitem()</code> removes the most recent pair (LIFO).
</div>

```python
d = {"a": 1, "b": 2, "c": 3}
print(d.pop("a"))
print(d)
print(d.popitem())
print(d)# removes last inserted (c)
del d["b"]
print(d)
d = {"a": 1, "b": 2, "c": 3}
d.clear()
print(d)
```

<div style="background:linear-gradient(90deg,#EFF6FF,#ECFDF5);border:1px solid #CBD5E1;border-radius:10px;padding:16px 20px;color:#1E293B;"><div style="font-size:13px;letter-spacing:2px;color:#3B82F6;font-weight:700;">SECTION IV</div><div style="font-size:22px;font-weight:800;margin-top:2px;">Iterating Dictionaries</div><div style="font-size:14px;color:#64748B;margin-top:6px;">Views over keys, values and pairs — and why they stay in sync with the dict.</div></div>

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Views & Iteration</span></div>

#### 1. keys(), values(), items()

**What it is** — These return view objects over the keys, the values, and the <code>(key, value)</code> pairs.

**Why it matters** — They are the standard way to loop over a dictionary’s contents.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Iterate pairs with <code>for k, v in d.items()</code>; default iteration yields keys.
</div>

```python
d = {"a": 1, "b": 2}
print(list(d.keys()))
print(list(d.values()))
for k, v in d.items():
    print(k, "->", v)
```

#### 2. Views Are Live

**What it is** — View objects reflect later changes to the dict — they are windows, not snapshots.

**Why it matters** — It explains why a saved <code>.keys()</code> view updates after you edit the dict.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
A view is a live window onto the dict; mutate the dict and the view shows the change.
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Views track the dict in real time — copy with <code>list()</code> for a frozen snapshot.
</div>

```python
d = {"a": 1}
ks = d.keys()
d["b"] = 2              # added after the view was taken
print(list(ks))
```

#### 3. Sorting While Iterating

**What it is** — Wrap items in <code>sorted()</code> with a <code>key</code> to iterate in a chosen order without changing the dict.

**Why it matters** — Reports and displays usually need a deterministic, sorted order.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>sorted(d.items(), key=...)</code> iterates in order while leaving the dict untouched.
</div>

```python
scores = {"Arun": 80, "Bina": 95, "Carl": 70}
for name, score in sorted(scores.items(), key=lambda kv: kv[1]):
    print(name, score)
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Rules & Guarantees</span></div>

#### 1. Keys Must Be Hashable

**What it is** — Dict keys must be hashable (typically immutable); values can be anything.

**Why it matters** — Hashing is how the dict locates entries in constant time.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Keys must be hashable (str, int, tuple…); a list key raises TypeError.
</div>

```python
d = {(1, 2): "ok", "name": "Arun",1:2}
print(d)

try:
    d[[1, 2]] = "bad"   # list key is unhashable
except TypeError as e:
    print("TypeError:", e)
```

#### 2. Guaranteed Insertion Order

**What it is** — Since Python 3.7, dictionaries remember the order in which keys were inserted.

**Why it matters** — You can rely on predictable iteration order for serialisation and display.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Dicts iterate in insertion order — a language guarantee since 3.7.
</div>

```python
d = {}
for key in ["z", "a", "m"]:
    d[key] = key.upper()
print(list(d))          # preserves z, a, m order
```

#### 3. Membership Tests the Keys

**What it is** — <code>key in d</code> checks keys, not values, and is an O(1) hash lookup.

**Why it matters** — It is the fast, idiomatic way to guard access to a possibly-missing key.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;<code>in</code> tests keys in constant time; use <code>in d.values()</code> to search values.
</div>

```python
d = {"a": 1, "b": 2}
print("a" in d)
print(1 in d)               # value, not key -> False
print(1 in d.values())
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Extended Mappings</span></div>

#### 1. collections.defaultdict

**What it is** — <code>defaultdict(factory)</code> auto-creates a missing key’s value by calling the factory (e.g. <code>list</code>, <code>int</code>).

**Why it matters** — It removes the boilerplate of checking for and initialising missing keys.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;defaultdict supplies a fresh default on first access — no key-existence checks.
</div>

```python
from collections import defaultdict
groups = defaultdict(list)
for name, team in [("Arun", "A"), ("Bina", "B"), ("Carl", "A")]:
    groups[team].append(name)
print(dict(groups))
```

#### 2. collections.Counter

**What it is** — <code>Counter</code> is a dict subclass that tallies hashable items and ranks them with <code>most_common()</code>.

**Why it matters** — It turns frequency counting — a constant need — into a single call.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Counter counts occurrences and ranks them, replacing hand-written tally loops.
</div>

```python
from collections import Counter
votes = ["a", "b", "a", "c", "a", "b"]
c = Counter(votes)
print(c)
print(c.most_common(2))
print(c["a"])
```

#### 3. OrderedDict (Mostly Legacy)

**What it is** — <code>OrderedDict</code> predates the 3.7 ordering guarantee; today it is needed only for order-sensitive equality and <code>move_to_end()</code>.

**Why it matters** — Knowing its niche prevents reaching for it when a plain dict already suffices.

<div style="background:#F8FAFC;border-left:5px solid #64748B;padding:10px;border-radius:6px;color:#1E293B;"><b>Important Note</b><br>
Since 3.7 a plain dict keeps insertion order — reach for OrderedDict only for <code>move_to_end()</code> or order-sensitive equality.
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Plain dicts are ordered now; OrderedDict survives for a few specialised behaviours.
</div>

```python
from collections import OrderedDict
od = OrderedDict([("a", 1), ("b", 2)])
od.move_to_end("a")
print(list(od))
```

<div style="border-left:4px solid #3B82F6;padding-left:12px;color:#1E293B;"><span style="font-size:18px;font-weight:700;">A. Choosing & Copying</span></div>

#### 1. When to Reach for a Dict

**What it is** — Use a dict when you look data up by a label. Membership and lookup in a dict are O(1) hash operations; the same search in a list is an O(n) scan.

**Why it matters** — For repeated lookups over large data this is the single biggest performance lever.

<div style="background:#EFF6FF;border-left:5px solid #3B82F6;padding:10px;border-radius:6px;color:#1E293B;"><b>Executive Insight</b><br>
<code>x in some_list</code> scans every element (O(n)); <code>key in some_dict</code> is a single hash lookup (O(1)).
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Reach for a dict whenever you answer ‘what is the value for this key?’ — it beats scanning a list decisively.
</div>

```python
big_list = list(range(100_000))
lookup = {n: True for n in big_list}
print(99_999 in big_list)   # scans up to 100k items
print(99_999 in lookup)     # one hash lookup
```

#### 2. Copy Safely: Shared, Shallow & Deep

**What it is** — <code>b = a</code> binds a second name to the same dict; <code>d.copy()</code> / <code>dict(d)</code> make a shallow copy (top level only); <code>copy.deepcopy()</code> clones nested objects too.

**Why it matters** — A shallow copy still shares the inner lists/dicts — only a deep copy is fully independent, which is the most common source of ‘why did my other dict change?’ bugs.

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:10px;border-radius:6px;color:#1E293B;"><b>Best Practice</b><br>
Use a shallow copy for flat data; reach for <code>copy.deepcopy()</code> only when the structure is nested and must be fully separate.
</div>

<div style="background:#ECFDF5;border-left:5px solid #10B981;padding:8px 10px;border-radius:6px;color:#1E293B;"><b>&#127919; Key Takeaway</b> &nbsp;Assignment shares; <code>.copy()</code>/<code>dict(d)</code> duplicates the container; <code>deepcopy()</code> duplicates everything inside it.
</div>

```python
import copy
original = {"nums": [1, 2, 3]}
alias = original            # same object
shallow = original.copy()   # new dict, shared inner list
deep = copy.deepcopy(original)
shallow["nums"].append(99)  # leaks into original
print("after shallow edit:", original["nums"])
deep["nums"].append(0)      # isolated
print("after deep edit   :", original["nums"])
print("alias is original :", alias is original)
```
